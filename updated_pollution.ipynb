Identifikasi dataset:

Jenis Klasifikasi: Multiclass Classification.
- Independent Variables: Temperature, Humidity, PM2.5, PM10, NO2, SO2, CO, dan Proximity_to_Industrial_Areas.
- Dependent Variable: Air Quality.
- Jumlah Independent Variables: 8 fitur.
- Jumlah Dependent Variables: 1 target.
- Jumlah Kategori Target: 4 kategori.
- Kategori: Good, Moderate, Poor, dan Hazardous.

Penjelasan variabel:

1. Temperature: Menunjukkan suhu lingkungan.
2. Humidity: Menunjukkan tingkat kelembapan udara.
3. PM2.5: Menunjukkan konsentrasi partikel halus berukuran hingga 2,5 mikrometer.
4. PM10: Menunjukkan konsentrasi partikel berukuran hingga 10 mikrometer.
5. NO2: Menunjukkan konsentrasi nitrogen dioksida.
6. SO2: Menunjukkan konsentrasi sulfur dioksida.
7. CO: Menunjukkan konsentrasi karbon monoksida.
8. Proximity_to_Industrial_Areas: Menunjukkan kedekatan lokasi dengan kawasan industri.

Penjelasan dataset: Air Quality and Pollution Assessment merupakan dataset yang digunakan untuk mengidentifikasi tingkat kualitas udara berdasarkan beberapa parameter lingkungan dan polusi.

Dataset ini menggambarkan kondisi udara melalui pengukuran seperti suhu, kelembapan, konsentrasi partikel polusi, dan kandungan gas tertentu. Hasil pengukuran tersebut digunakan untuk menentukan tingkat kualitas udara.

Dataset ini cocok untuk mempelajari klasifikasi multiclass karena hasil akhirnya terdiri dari beberapa kategori kualitas udara, bukan hanya dua kategori.
